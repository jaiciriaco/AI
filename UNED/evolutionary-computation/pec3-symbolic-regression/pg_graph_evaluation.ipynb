# Gráficas PEC3

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df_tuning = pd.read_csv('pg_resumen_tuning.csv')
df_final = pd.read_csv('pg_resumen_final.csv')
df_hist_final = pd.read_csv('pg_historial_final.csv')
df_pred = pd.read_csv('pg_predicciones_final.csv')

df_tuning.head()
```

## Resumen tuning problema 1

```python
df_tuning.sort_values(['TE', 'VAMM', 'PEX'], ascending=[False, True, True])
```

## VAMM por problema

```python
plt.figure(figsize=(10,6))
for problema in sorted(df_final['problema'].unique()):
    sub = df_final[df_final['problema'] == problema]
    plt.bar(problema, sub['VAMM'].mean(), yerr=sub['DE_VAMM'].mean(), capsize=4)
plt.ylabel('VAMM ± DE')
plt.title('VAMM por problema')
plt.grid(True, axis='y')
plt.show()
```

## TE por problema

```python
plt.figure(figsize=(10,6))
for problema in sorted(df_final['problema'].unique()):
    sub = df_final[df_final['problema'] == problema]
    plt.bar(problema, sub['TE'].mean())
plt.ylabel('TE')
plt.title('Tasa de éxito por problema')
plt.ylim(0, 1.05)
plt.grid(True, axis='y')
plt.show()
```

## PEX por problema

```python
plt.figure(figsize=(10,6))
for problema in sorted(df_final['problema'].unique()):
    sub = df_final[df_final['problema'] == problema]
    valor = sub['PEX'].mean()
    error = sub['DE_PEX'].mean()
    plt.bar(problema, valor, yerr=error, capsize=4)
plt.ylabel('PEX ± DE')
plt.title('Promedio de evaluaciones para alcanzar el éxito')
plt.grid(True, axis='y')
plt.show()
```

## Curvas de progreso por problema

```python
for problema in sorted(df_hist_final['problema'].unique()):
    plt.figure(figsize=(10,6))
    sub = df_hist_final[df_hist_final['problema'] == problema].copy()
    tabla = sub.pivot_table(index='generacion', columns='ejecucion', values='fitness', aggfunc='last')
    media = tabla.mean(axis=1)
    desviacion = tabla.std(axis=1)
    plt.plot(media.index, media.values, label='media')
    plt.fill_between(media.index, (media - desviacion).values, (media + desviacion).values, alpha=0.2)
    plt.xlabel('Generación')
    plt.ylabel('Fitness medio')
    plt.title(f'Curva de progreso - {problema}')
    plt.grid(True)
    plt.legend()
    plt.show()
```

## Función real vs mejor aproximación

```python
for problema in sorted(df_pred['problema'].unique()):
    plt.figure(figsize=(10,6))
    sub = df_pred[df_pred['problema'] == problema].copy()
    plt.plot(sub['x'], sub['y_real'], label='f(x) real')
    plt.plot(sub['x'], sub['y_pred'], label='f_hat(x) PG')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title(f'Función real vs aproximación - {problema}')
    plt.grid(True)
    plt.legend()
    plt.show()
```

## Error absoluto

```python
for problema in sorted(df_pred['problema'].unique()):
    plt.figure(figsize=(10,6))
    sub = df_pred[df_pred['problema'] == problema].copy()
    error_abs = (sub['y_real'] - sub['y_pred']).abs()
    plt.plot(sub['x'], error_abs)
    plt.yscale('log')
    plt.xlabel('x')
    plt.ylabel('Error absoluto (log)')
    plt.title(f'Error absoluto - {problema}')
    plt.grid(True)
    plt.show()
```
