## 1–2. Probabilidad de eventos



### Ejercicio 1 — Solución


```python

import numpy as np

rng = np.random.default_rng(0)
n = 10000
tiradas = rng.integers(1, 7, size=n)

prob_mayor_4 = np.mean(tiradas > 4)      # {5,6}
prob_impar = np.mean(tiradas % 2 == 1)   # {1,3,5}

print("P(resultado > 4) ≈", prob_mayor_4, " (teórico: 2/6 ≈ 0.333)")
print("P(resultado impar) ≈", prob_impar, " (teórico: 3/6 = 0.5)")
```


## 3. Probabilidad condicional


```python

visitas_total = 1000
movil = 400
pc = 600

clic_movil = 80
clic_pc = 60

p_movil = movil / visitas_total
p_pc = pc / visitas_total

p_clic = (clic_movil + clic_pc) / visitas_total

p_clic_dado_movil = clic_movil / movil
p_clic_dado_pc = clic_pc / pc

print("P(móvil) =", p_movil)
print("P(PC) =", p_pc)
print("P(clic) =", p_clic)
print("P(clic | móvil) =", p_clic_dado_movil)
print("P(clic | PC) =", p_clic_dado_pc)
```


### Ejercicio 2 — Solución


```python

# P(móvil y clic) = clic_movil / visitas_total
p_mov_y_clic = clic_movil / visitas_total
print("P(móvil ∩ clic) =", p_mov_y_clic)

# Verificación usando P(clic|móvil) * P(móvil)
print("P(clic|móvil)*P(móvil) =", p_clic_dado_movil * p_movil)
```


## 4. Teorema de Bayes


```python

sens = 0.99
spec = 0.95
prev = 0.01  # prevalencia 1%

p_pos = sens * prev + (1 - spec) * (1 - prev)
p_enf_dado_pos = sens * prev / p_pos

print("Con prevalencia 1%:")
print("P(positivo) =", p_pos)
print("P(enfermo | positivo) ≈", p_enf_dado_pos)
```


### Ejercicio 3 — Solución


```python

sens = 0.99
spec = 0.95
prev = 0.10  # prevalencia 10%

p_pos = sens * prev + (1 - spec) * (1 - prev)
p_enf_dado_pos = sens * prev / p_pos

print("Con prevalencia 10%:")
print("P(positivo) =", p_pos)
print("P(enfermo | positivo) ≈", p_enf_dado_pos)
```


## 5. Variables aleatorias y distribuciones


```python

import matplotlib.pyplot as plt

rng = np.random.default_rng(1)
muestra = rng.normal(loc=5, scale=2, size=10000)

plt.figure()
plt.hist(muestra, bins=30)
plt.title("Distribución normal simulada (media 5, sd 2)")
plt.xlabel("valor"); plt.ylabel("frecuencia")
plt.show()
```


## 6. Ley de los Grandes Números


```python

rng = np.random.default_rng(42)
p = 0.6
n = 2000
caras_acum = np.cumsum(rng.random(n) < p)
proporcion = caras_acum / np.arange(1, n+1)

plt.figure()
plt.plot(proporcion)
plt.axhline(p)
plt.title("Ley de los Grandes Números (moneda sesgada p=0.6)")
plt.xlabel("número de lanzamientos"); plt.ylabel("proporción de caras")
plt.show()
```


### Ejercicio 5 — Solución ejemplo


```python

for p in [0.2, 0.8]:
    rng = np.random.default_rng(0)
    n = 200
    caras_acum = np.cumsum(rng.random(n) < p)
    proporcion = caras_acum / np.arange(1, n+1)
    print(f"\nSimulación con p={p}: última proporción ≈ {proporcion[-1]:.3f}")
```
