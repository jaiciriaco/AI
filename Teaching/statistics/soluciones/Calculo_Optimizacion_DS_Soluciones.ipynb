## 1. Funciones y pendiente


```python

import numpy as np
import matplotlib.pyplot as plt

def f_lineal(x):
    return 2*x + 1

xs = np.linspace(-5, 5, 100)
ys = f_lineal(xs)

plt.figure()
plt.plot(xs, ys)
plt.title("Función lineal f(x) = 2x + 1")
plt.xlabel("x"); plt.ylabel("f(x)")
plt.show()
```


### Ejercicio 1 — Solución


```python

import numpy as np
import matplotlib.pyplot as plt

def f_lineal_pos(x):
    return 0.5 * x + 1  # pendiente suave, positiva

def f_lineal_neg(x):
    return -1 * x + 1   # pendiente negativa

xs = np.linspace(-5, 5, 100)

plt.figure()
plt.plot(xs, f_lineal_pos(xs))
plt.title("Pendiente 0.5 (crece lento)")
plt.xlabel("x"); plt.ylabel("f(x)")
plt.show()

plt.figure()
plt.plot(xs, f_lineal_neg(xs))
plt.title("Pendiente -1 (decrece)")
plt.xlabel("x"); plt.ylabel("f(x)")
plt.show()
```


## 2. Derivada: qué es y qué significa


```python

def f_quad(x):
    return x**2

xs = np.linspace(-4, 4, 200)
ys = f_quad(xs)

plt.figure()
plt.plot(xs, ys)
plt.title("Función cuadrática f(x) = x^2")
plt.xlabel("x"); plt.ylabel("f(x)")
plt.show()
```


### Ejercicio 2 — Solución (comentada)

- La función decrece para x < 0 y crece para x > 0.  
- En x = 0 la pendiente es 0: es el mínimo de la parábola.


```python

# Podemos comprobar numéricamente la pendiente alrededor de 0
def approx_derivative(f, x, h=1e-3):
    return (f(x + h) - f(x - h)) / (2*h)

for point in [-2, -1, 0, 1, 2]:
    print(f"Derivada aproximada en x={point}: {approx_derivative(f_quad, point):.3f}")
```


## 3. Mínimos, máximos y descenso del gradiente en 1D


```python

def f(w):
    return (w - 3)**2 + 1

def df(w):
    return 2*(w - 3)

w0 = -5.0
lr = 0.1
hist = []
w = w0
for i in range(40):
    hist.append(w)
    w = w - lr * df(w)

print("Valor aproximado de w en el mínimo:", w)

ws_plot = np.linspace(-6, 8, 200)
fs_plot = f(ws_plot)

plt.figure()
plt.plot(ws_plot, fs_plot)
plt.scatter(hist, [f(v) for v in hist])
plt.title("Descenso del gradiente en f(w) = (w-3)^2 + 1")
plt.xlabel("w"); plt.ylabel("f(w)")
plt.show()
```


### Ejercicio 3 — Solución (comparación de learning rates)


```python

def run_gd(lr, steps=20):
    def f(w):
        return (w - 3)**2 + 1
    def df(w):
        return 2*(w - 3)
    w = -5.0
    for i in range(steps):
        w = w - lr * df(w)
    return w, f(w)

for lr in [0.01, 0.1, 0.5, 1.5]:
    w_final, f_final = run_gd(lr, steps=40)
    print(f"lr={lr}: w_final={w_final:.3f}, f(w_final)={f_final:.3f}")
```


## 4. Descenso del gradiente en regresión lineal simple


```python

rng = np.random.default_rng(0)
x = np.linspace(0, 10, 50)
true_a, true_b = 2.0, 1.0
y = true_a * x + true_b + rng.normal(0, 2.0, size=x.size)

def mse(a, b):
    y_hat = a*x + b
    return np.mean((y - y_hat)**2)

def grad_mse(a, b):
    y_hat = a*x + b
    da = (-2/len(x)) * np.sum(x * (y - y_hat))
    db = (-2/len(x)) * np.sum(     (y - y_hat))
    return da, db

a, b = -1.0, 5.0
lr = 0.01
for _ in range(1000):
    da, db = grad_mse(a, b)
    a -= lr * da
    b -= lr * db

print(f"Valores encontrados: a≈{a:.3f}, b≈{b:.3f}")
print(f"Valores verdaderos: a={true_a}, b={true_b}")
print("MSE final:", mse(a, b))

plt.figure()
plt.scatter(x, y)
plt.plot(x, a*x + b)
plt.title("Ajuste lineal por descenso del gradiente")
plt.xlabel("x"); plt.ylabel("y")
plt.show()
```


### Ejercicio 4 — Solución ejemplo


```python

# Probamos distinto punto inicial y menos iteraciones

a, b = 0.0, 0.0
lr = 0.01
steps = 200
for _ in range(steps):
    da, db = grad_mse(a, b)
    a -= lr * da
    b -= lr * db

print(f"Tras {steps} pasos: a≈{a:.3f}, b≈{b:.3f}, MSE≈{mse(a,b):.3f}")
```
