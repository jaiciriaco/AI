## 1. Media y mediana de salarios


```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(0)
salarios = np.concatenate([
    rng.normal(25, 5, size=90),
    rng.normal(60, 10, size=10)
])
salarios = salarios.clip(10, None)

serie = pd.Series(salarios)
print("Media:", serie.mean())
print("Mediana:", serie.median())
print("Mínimo:", serie.min())
print("Máximo:", serie.max())
```


### Ejercicio 1 — Solución


```python

rng = np.random.default_rng(0)
salarios2 = np.concatenate([
    rng.normal(25, 5, size=50),   # menos salarios "normales"
    rng.normal(60, 10, size=50)   # muchos salarios altos
])
salarios2 = salarios2.clip(10, None)
serie2 = pd.Series(salarios2)

print("Media nueva:", serie2.mean())
print("Mediana nueva:", serie2.median())
```


## 2. Dispersión: varianza y desviación estándar


## 4. Intervalo de confianza de la media


```python

m = serie.mean()
s = serie.std(ddof=1)
n = len(serie)
se = s / np.sqrt(n)
z = 1.96
ic_low, ic_high = m - z*se, m + z*se

print(f"Media muestral ≈ {m:.2f}")
print(f"IC 95% de la media ≈ [{ic_low:.2f}, {ic_high:.2f}]")
```


### Ejercicio 4 — Solución


```python

sub = serie.sample(30, random_state=123)
m_sub = sub.mean()
s_sub = sub.std(ddof=1)
n_sub = len(sub)
se_sub = s_sub / np.sqrt(n_sub)
z = 1.96
ic_low_sub, ic_high_sub = m_sub - z*se_sub, m_sub + z*se_sub

print(f"Media submuestra ≈ {m_sub:.2f}")
print(f"IC 95% submuestra ≈ [{ic_low_sub:.2f}, {ic_high_sub:.2f}]")
```


## 5. Correlación horas de estudio vs nota


```python

rng = np.random.default_rng(123)
N = 200
horas_estudio = rng.normal(4, 1.5, size=N).clip(0, None)
ruido = rng.normal(0, 1.5, size=N)
nota = 2.5*horas_estudio + 3 + ruido

df = pd.DataFrame({"horas_estudio": horas_estudio, "nota": nota})

corr = df["horas_estudio"].corr(df["nota"])
print("Correlación Pearson horas_estudio vs nota =", round(corr, 3))

plt.figure()
plt.scatter(df["horas_estudio"], df["nota"])
plt.title("Relación horas de estudio vs nota")
plt.xlabel("horas de estudio"); plt.ylabel("nota")
plt.show()
```


### Ejercicio 5 — Solución


```python

ruido_fuerte = ruido * 3
nota2 = 2.5*horas_estudio + 3 + ruido_fuerte

df2 = pd.DataFrame({"horas_estudio": horas_estudio, "nota2": nota2})
corr2 = df2["horas_estudio"].corr(df2["nota2"])

print("Correlación con más ruido:", round(corr2, 3))
```


## 6. Regresión lineal simple (OLS)


```python

X = np.c_[np.ones(len(df)), df["horas_estudio"].values]
y = df["nota"].values.reshape(-1, 1)

w = np.linalg.lstsq(X, y, rcond=None)[0]
b_hat, a_hat = float(w[0, 0]), float(w[1, 0])
print(f"Modelo estimado: nota ≈ {a_hat:.3f} * horas + {b_hat:.3f}")

y_hat = (X @ w).ravel()
mse = np.mean((df["nota"].values - y_hat)**2)
print("MSE OLS:", round(mse, 3))

plt.figure()
plt.scatter(df["horas_estudio"], df["nota"])
x_line = np.linspace(df["horas_estudio"].min(), df["horas_estudio"].max(), 100)
y_line = a_hat * x_line + b_hat
plt.plot(x_line, y_line)
plt.title("Regresión lineal simple (OLS)")
plt.xlabel("horas de estudio"); plt.ylabel("nota")
plt.show()
```


### Ejercicio 6 — Solución ejemplo


```python

# Añadimos dos outliers extremos
df_out = df.copy()
df_out.loc[len(df_out)] = [0, 10]   # 0 horas, nota 10
df_out.loc[len(df_out)] = [10, 0]   # 10 horas, nota 0

X_out = np.c_[np.ones(len(df_out)), df_out["horas_estudio"].values]
y_out = df_out["nota"].values.reshape(-1, 1)

w_out = np.linalg.lstsq(X_out, y_out, rcond=None)[0]
b_hat_out, a_hat_out = float(w_out[0, 0]), float(w_out[1, 0])

corr_out = df_out["horas_estudio"].corr(df_out["nota"])

print(f"Nuevo modelo con outliers: nota ≈ {a_hat_out:.3f} * horas + {b_hat_out:.3f}")
print("Nueva correlación con outliers:", round(corr_out, 3))
```
