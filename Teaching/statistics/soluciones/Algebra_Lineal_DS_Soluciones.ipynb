
## 2. Normas y distancia


```python

x = np.array([3, 4])
y = np.array([0, 0])

norm_x = np.linalg.norm(x)
dist_xy = np.linalg.norm(x - y)

print("x =", x)
print("Norma de x =", norm_x)
print("Distancia entre x e y =", dist_xy)
```


### Ejercicio 2 — Solución


```python

import numpy as np

a = np.array([1, 2, 3])
b = np.array([2, 0, 1])

norm_a = np.linalg.norm(a)
norm_b = np.linalg.norm(b)
dist_ab = np.linalg.norm(a - b)

print("a =", a, "| ||a|| =", norm_a)
print("b =", b, "| ||b|| =", norm_b)
print("Distancia entre a y b =", dist_ab)
```


## 3. Producto escalar y similitud coseno


```python

x = np.array([1, 2, 3])
y = np.array([4, 0, -1])

dot = np.dot(x, y)
norm_x = np.linalg.norm(x)
norm_y = np.linalg.norm(y)
cos_sim = dot / (norm_x * norm_y)

print("x =", x)
print("y =", y)
print("Producto escalar x·y =", dot)
print("Similitud coseno =", cos_sim)
```


### Ejercicio 3 — Solución


```python

usuarios = {
    "A": np.array([5, 1, 2]),
    "B": np.array([4, 2, 1]),
    "C": np.array([1, 5, 4])
}

A = usuarios["A"]

def cos_sim(u, v):
    return np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))

for nombre, v in usuarios.items():
    if nombre == "A":
        continue
    cs = cos_sim(A, v)
    print(f"Similitud coseno A vs {nombre} = {cs:.3f}")
```


## 5. Operaciones con matrices y regresión lineal simple


```python

X = np.array([
    [1, 1.0],
    [1, 2.0],
    [1, 3.0],
    [1, 4.0]
])
y = np.array([5, 6, 7, 8])

w = np.array([4, 1])  # y_hat = 4 + 1*horas

y_hat = X @ w
print("X =\n", X)
print("w =", w)
print("Predicciones y_hat =", y_hat)
```


### Ejercicio 5 — Solución


```python

X = np.array([
    [1, 1.0],
    [1, 2.0],
    [1, 3.0],
    [1, 4.0]
])
y = np.array([5, 6, 7, 8])

# Ejemplo 1: sin término independiente (bajo)
w1 = np.array([0, 2])
y_hat1 = X @ w1

# Ejemplo 2: término independiente alto y poca pendiente
w2 = np.array([3, 0.5])
y_hat2 = X @ w2

print("w1 =", w1, "| y_hat1 =", y_hat1)
print("w2 =", w2, "| y_hat2 =", y_hat2)
```
